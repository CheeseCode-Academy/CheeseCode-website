اگر بخواهیم یک فایل html بنویسیم باید استایل کلی نوشتن ما به صورت زیر باشه:

````python
<!DOCTYPE html>
<html>
    <head>
        <!--your code-->
    </head>
    <body>
        <!--your code-->
    </body>
</html>
````


در بالا میشود که تگ زیر نوشته نشود اما بهتر است که بنویسیم:
```html
<!DOCTYPE html>
```
تگ بالا به برنامه مرورگر میفهماند که باید از نسخه جدید استفاده کند

توضیحات مابقی تگ های به کار رفته:
```html
<html>
    <!-- ... -->
</html>
```
این تگ صفحه اصلی را میسازد و تمام محتوا و تگ ها بین این دو قرار میگیرند

تگ <head>:
```html
<head>
    <!--your code-->
</head>
```
این تگ برای لود اولیه فایل های وابسته به این کار میکند و همینطور برای اطلاعات متا دیتا

تگ <body>:
```html
<body>
    <!--your code-->
</body>
```
در بین این دو تمام محتوای سایت نوشته میشود و این بدنه سایتی است که میسازیم

**نکات:**
1. یادمان نرود که تگ هایی که باز کردیم را ببندیم
2. بعضی از تگ ها نیاز به بسته شدن ندارند که در مثال ها و نوت ها هم آنها را نمیبندیم
